```python
import pandas as pd
import numpy as np
from itertools import combinations

# Labels und ihre semantischen Beziehungen
labels = [
    "101 - Foreign Special Relationships: Positive",
    "102 - Foreign Special Relationships: Negative",
    "103 - Anti-Imperialism",
    "104 - Military: Positive",
    "105 - Military: Negative",
    "106 - Peace",
    "107 - Internationalism: Positive",
    "108 - European Community/Union: Positive",
    "109 - Internationalism: Negative",
    "110 - European Community/Union: Negative",
    "201 - Freedom and Human Rights",
    "202 - Democracy",
    "203 - Constitutionalism: Positive",
    "204 - Constitutionalism: Negative",
    "301 - Decentralization",
    "302 - Centralisation: Positive",
    "303 - Governmental and Administrative Efficiency",
    "304 - Political Corruption",
    "305 - Political Authority",
    "401 - Free Market Economy",
    "402 - Incentives: Positive",
    "403 - Market Regulation",
    "404 - Economic Planning",
    "405 - Corporatism/Mixed Economy",
    "406 - Protectionism: Positive",
    "407 - Protectionism: Negative",
    "408 - Economic Goals",
    "409 - Keynesian Demand Management",
    "410 - Economic Growth: Positive",
    "411 - Technology and Infrastructure: Positive",
    "412 - Controlled Economy",
    "413 - Nationalisation",
    "414 - Economic Orthodoxy",
    "415 - Marxist Analysis",
    "416 - Anti-Growth Economy and Sustainability",
    "501 - Environmental Protection: Positive",
    "502 - Culture: Positive",
    "503 - Equality: Positive",
    "504 - Welfare State Expansion",
    "505 - Welfare State Limitation",
    "506 - Education Expansion",
    "507 - Education Limitation",
    "601 - National Way of Life: Positive",
    "602 - National Way of Life: Negative",
    "603 - Traditional Morality: Positive",
    "604 - Traditional Morality: Negative",
    "605 - Law and Order: Positive",
    "606 - Civic Mindedness: Positive",
    "607 - Multiculturalism: Positive",
    "608 - Multiculturalism: Negative",
    "701 - Labour Groups: Positive",
    "702 - Labour Groups: Negative",
    "703 - Agriculture and Farmers: Positive",
    "704 - Middle Class and Professional Groups",
    "705 - Underprivileged Minority Groups",
    "706 - Non-economic Demographic Groups",
]

# Gegenteilige Paare (ohne explizit Positive/Negative im Namen)
opposites = [
    ("101 - Foreign Special Relationships: Positive", "102 - Foreign Special Relationships: Negative"),
    ("104 - Military: Positive", "105 - Military: Negative"),
    ("107 - Internationalism: Positive", "109 - Internationalism: Negative"),
    ("108 - European Community/Union: Positive", "110 - European Community/Union: Negative"),
    ("203 - Constitutionalism: Positive", "204 - Constitutionalism: Negative"),
    ("301 - Decentralization", "302 - Centralisation: Positive"),
    ("401 - Free Market Economy", "404 - Economic Planning"),
    ("402 - Incentives: Positive", "409 - Keynesian Demand Management"),
    ("406 - Protectionism: Positive", "407 - Protectionism: Negative"),
    ("504 - Welfare State Expansion", "505 - Welfare State Limitation"),
    ("506 - Education Expansion", "507 - Education Limitation"),
    ("601 - National Way of Life: Positive", "602 - National Way of Life: Negative"),
    ("603 - Traditional Morality: Positive", "604 - Traditional Morality: Negative"),
    ("607 - Multiculturalism: Positive", "608 - Multiculturalism: Negative"),
    ("701 - Labour Groups: Positive", "702 - Labour Groups: Negative"),
    ("704 - Middle Class and Professional Groups","705 - Underprivileged Minority Groups"),
    ("416 - Anti-Growth Economy and Sustainability", "410 - Economic Growth: Positive"),
    ("412 - Controlled Economy", "401 - Free Market Economy"),
    ("414 - Economic Orthodoxy", "409 - Keynesian Demand Management"),
]

# Semantisch ähnliche Gruppen
similar_groups = [
    {"101 - Foreign Special Relationships: Positive", "107 - Internationalism: Positive"},
    {"102 - Foreign Special Relationships: Negative", "109 - Internationalism: Negative"},
    {"401 - Free Market Economy", "402 - Incentives: Positive"},
    {"501 - Environmental Protection: Positive", "416 - Anti-Growth Economy and Sustainability"},
    {"601 - National Way of Life: Positive","603 - Traditional Morality: Positive", "605 - Law and Order: Positive"},
    {"504 - Welfare State Expansion", "506 - Education Expansion"},
    {"505 - Welfare State Limitation", "507 - Education Limitation"},
]

# Adjazenzmatrix initialisieren
matrix = pd.DataFrame(0, index=labels, columns=labels)

# Gegenteilige Labels mit -1 belegen
for label1, label2 in opposites:
    matrix.at[label1, label2] = -1
    matrix.at[label2, label1] = -1

# Semantisch ähnliche Gruppen mit 1 belegen
for group in similar_groups:
    for label1, label2 in combinations(group, 2):
        matrix.at[label1, label2] = 1
        matrix.at[label2, label1] = 1

for label in labels:
    matrix.at[label, label] = 1
```
